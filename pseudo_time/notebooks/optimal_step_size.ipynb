```python
import sympy as sym
```

```python
alpha, beta, eta_m = sym.symbols(r'alpha beta eta_m')
tau = sym.symbols('tau')
```

```python
eta = sym.exp(- alpha * tau) / eta_m + beta * tau * sym.sqrt(1 - sym.exp(-0.5 * eta_m * tau))
eta
```

```python
print(eta.diff(tau))
```
